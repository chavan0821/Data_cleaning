```python
# stadarize gender
df['Gender'] = df['Gender'].str.title()
df['Gender'].unique()
```

```python
# check numeic columns.
numeric_columns = [
    "Customer_ID",
    "Age",
    "Monthly_Income",
    "Latitude",
    "Longitude",
    "Purchase_Amount"
]

df[numeric_columns].dtypes
```

```python
# Handel missing age 

df['Age'].isnull().sum()
```

```python
# fill missing age using the median.

df['Age']=df['Age'].fillna(df["Age"].median())

df["Age"].isnull().sum()
```

```python
# Handle missing monthly income.

df['Monthly_Income']=df['Monthly_Income'].fillna(
    df["Monthly_Income"].median())

```

```python
# Handel missing latitude.
df["Latitude"] = df["Latitude"].fillna(
    df["Latitude"].median()
)
```

```python
# Handle missing longitude.
df["Longitude"] = df["Longitude"].fillna(
    df["Longitude"].median()
)
```

```python
# handle missing purchasee amount 

df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].median()
)
```

```python
# Handle missing Categorical values.
# Gender
df["Gender"]= df['Gender'].fillna(
    df['Gender'].mode()[0])
```

```python
# Department
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)
```

```python
# City
df["City"] = df["City"].fillna(
    df["City"].mode()[0]
)
```

```python
#Customer Type
df["Customer_Type"] = df["Customer_Type"].fillna(
    df["Customer_Type"].mode()[0]
)
```

```python
# Feedback.
df["Feedback"]=df["Feedback"].fillna(
    df["Feedback"].mode()[0])
```

```python
# Handel missing name
df["Name"] = df["Name"].fillna("Unknown")
```

```python
# Final missing value check.
df.isnull().sum()
```
